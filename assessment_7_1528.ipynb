```python
def calculate_returns(rewards, gamma):
    """
    Calculates the undiscounted and discounted returns for a sequence of rewards.

    Args:
        rewards (list): A sequence of reward values.
        gamma (float): The discount factor.

    Returns:
        tuple: (undiscounted_return, discounted_return)
    """
    undiscounted_return = sum(rewards)

    discounted_return = 0.0
    for t, r in enumerate(rewards):
        discounted_return += (gamma ** t) * r

    return undiscounted_return, discounted_return

def run_interactive_calculator():
    """
    A helper function to run the return calculator interactively.
    """
    try:
        steps = int(input("Enter the number of time steps/episodes: "))
        rewards = []
        for i in range(steps):
            r = float(input(f"Enter the reward for time step {i+1}: "))
            rewards.append(r)

        gamma = float(input("Enter the discount factor gamma (0 to 1): "))
        if not (0 <= gamma <= 1):
            print("Warning: Gamma typically ranges between 0 and 1.")

        undiscounted, discounted = calculate_returns(rewards, gamma)

        print("\n--- Results ---")
        print(f"Entered reward values: {rewards}")
        print(f"Discount factor gamma: {gamma}")
        print(f"Undiscounted return:   {undiscounted}")
        print(f"Discounted return:     {discounted:.4f}")
    except ValueError:
        print("Invalid input. Please enter numeric values.")

# Example run with sample inputs:
sample_rewards = [1.0, 2.0, -1.0, 5.0]
sample_gamma = 0.9
undiscounted, discounted = calculate_returns(sample_rewards, sample_gamma)

print("Example Run Demonstration:")
print(f"Rewards: {sample_rewards}")
print(f"Gamma: {sample_gamma}")
print(f"Undiscounted Return: {undiscounted}")
print(f"Discounted Return: {discounted:.4f}")
```

## Assignment 2: 2x3 Grid World MDP Representation

```python
class GridWorldMDP:
    def __init__(self):
        # a. Define State Space
        # Let's represent the grid as:
        # (0,0) (0,1) (0,2)
        # (1,0) (1,1) (1,2)
        # Where (1,1) is an obstacle/wall, and (0,2) is the Goal state.
        self.states = [(r, c) for r in range(2) for c in range(3)]
        self.obstacle = (1, 1)
        self.goal = (0, 2)

        # b. Define Action Space
        # 'U' (Up), 'D' (Down), 'L' (Left), 'R' (Right)
        self.actions = ['U', 'D', 'L', 'R']

        # d. Define Discount Factor
        self.gamma = 0.9

    # c. Define Reward Function
    def get_reward(self, state):
        if state == self.goal:
            return 10.0
        elif state == self.obstacle:
            return -10.0  # Big penalty for entering the obstacle
        else:
            return -1.0   # Step cost to encourage shortest path

    # e. Define Transition Function & f. Estimate next state
    def step(self, state, action):
        """
        Determines the next state given the current state and action.
        The transition is deterministic. Moving into a wall or the boundary
        keeps the agent in its current state.
        """
        if state == self.goal:
            return state  # Goal is absorbing

        r, c = state

        if action == 'U':
            next_state = (r - 1, c)
        elif action == 'D':
            next_state = (r + 1, c)
        elif action == 'L':
            next_state = (r, c - 1)
        elif action == 'R':
            next_state = (r, c + 1)
        else:
            next_state = state

        # Check boundary bounds and obstacle collision
        if next_state not in self.states or next_state == self.obstacle:
            return state

        return next_state

# Create the environment instance
grid_env_det = GridWorldMDP()

# Demonstrate next-state transitions
test_transitions = [
    ((0, 0), 'R'),  # Moves to (0,1)
    ((0, 1), 'D'),  # Attempts to move into obstacle (1,1) -> stays at (0,1)
    ((0, 1), 'R'),  # Moves to Goal (0,2)
    ((0, 2), 'D'),  # Goal is absorbing -> stays at (0,2)
]

print("Deterministic MDP Transition Test:")
for s, a in test_transitions:
    next_s = grid_env_det.step(s, a)
    r = grid_env_det.get_reward(next_s)
    print(f"State {s} --Action {a}--> Next State: {next_s} | Reward: {r}")
```
