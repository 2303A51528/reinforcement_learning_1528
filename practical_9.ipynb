```python
import numpy as np
import pandas as pd

print("--- Running Assignment 1: Algebraic Solution ---")

# Coefficients Matrix for system:
#  1 * V(S0) - 0.81 * V(S1) = -1.0
# -0.09 * V(S0) + 1 * V(S1) = 89.9

A = np.array([
    [1.0, -0.81],
    [-0.09, 1.0]
])
b = np.array([-1.0, 89.9])

# Solve the system directly
V_S0, V_S1 = np.linalg.solve(A, b)

# Calculate Action Values
gamma = 0.9
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1.0
Q_S1_Right = 100.0
Q_S1_Up = -1 + gamma * V_S0

# Verification check
V_S0_verify = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V_S1_verify = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

# Build Summary Table
summary_data_1 = {
    "State/Action": ["V_pi(S0)", "V_pi(S1)", "Q_pi(S0, Down)", "Q_pi(S0, Right)", "Q_pi(S1, Right)", "Q_pi(S1, Up)"],
    "Calculated Value": [round(V_S0, 2), round(V_S1, 2), round(Q_S0_Down, 2), round(Q_S0_Right, 2), round(Q_S1_Right, 2), round(Q_S1_Up, 2)]
}
df_summary_1 = pd.DataFrame(summary_data_1)

display(df_summary_1)

print(f"Verification status:")
print(f"V(S0) difference: {abs(V_S0 - V_S0_verify):.6f}")
print(f"V(S1) difference: {abs(V_S1 - V_S1_verify):.6f}")
```

```python
print("--- Running Assignment 2: Iterative Policy Evaluation ---")

# Parameters
gamma = 0.9
theta = 0.000001

# Initialize State Values
V_S0 = 0.0
V_S1 = 0.0

iterations = 0
while True:
    V_S0_old = V_S0
    V_S1_old = V_S1

    # Bellman update step
    V_S0_new = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V_S1_old)
    V_S1_new = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V_S0_old)

    delta = max(abs(V_S0_new - V_S0_old), abs(V_S1_new - V_S1_old))

    V_S0 = V_S0_new
    V_S1 = V_S1_new
    iterations += 1

    if delta < theta:
        break

# Compute Action Values (Q-values) using converged state values
Q_S0_Down = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V_S1) # Policy action
Q_S0_Right = 1.0 * (-1 + gamma * 0)                             # Alternative action
Q_S1_Right = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V_S0) # Policy action
Q_S1_Up = 1.0 * (-1 + gamma * V_S0)                             # Alternative action

# Build Summary Table
summary_data_2 = {
    "State/Action": [
        "Number of evaluation iterations",
        "V_pi(S0)",
        "V_pi(S1)",
        "Q_pi(S0, Down)",
        "Q_pi(S0, Right)",
        "Q_pi(S1, Right)",
        "Q_pi(S1, Up)"
    ],
    "Calculated Value": [
        float(iterations),
        round(V_S0, 6),
        round(V_S1, 6),
        round(Q_S0_Down, 6),
        round(Q_S0_Right, 6),
        round(Q_S1_Right, 6),
        round(Q_S1_Up, 6)
    ]
}
df_summary_2 = pd.DataFrame(summary_data_2)
display(df_summary_2)

# Consistency Check
print("Policy Consistency Checks:")
print(f"V_pi(S0) matches Q_pi(S0, Down): {np.isclose(V_S0, Q_S0_Down)}")
print(f"V_pi(S1) matches Q_pi(S1, Right): {np.isclose(V_S1, Q_S1_Right)}")
```
