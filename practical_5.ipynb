```python
import gymnasium as gym


def run_random_agent(env_name, episodes=5):
    print("\n" + "=" * 50)
    print("Environment:", env_name)
    print("=" * 50)

    env = gym.make(env_name)

    episode_rewards = []

    for episode in range(1, episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        terminated = False
        truncated = False

        print(f"\nEpisode {episode}")
        print("Initial State:", state)

        while not (terminated or truncated):

            # Select a random action
            action = env.action_space.sample()

            # Take action
            next_state, reward, terminated, truncated, info = env.step(action)

            print(
                f"Action: {action}, "
                f"Reward: {reward}, "
                f"Next State: {next_state}"
            )

            # Update state and reward
            state = next_state
            total_reward += reward

        episode_rewards.append(total_reward)

        print("Total Reward:", total_reward)

    # Calculate average reward
    average_reward = sum(episode_rewards) / len(episode_rewards)

    print("\n" + "-" * 50)
    print("Results for", env_name)
    print("-" * 50)
    print("Episode Rewards:", episode_rewards)
    print("Average Reward:", round(average_reward, 2))

    env.close()


# 1. CartPole
run_random_agent("CartPole-v1", episodes=5)

# 2. MountainCar
run_random_agent("MountainCar-v0", episodes=5)

# 3. FrozenLake
run_random_agent("FrozenLake-v1", episodes=5)
```
